This notebook trains the **OptimusPrime** transformer model to learn mappings from MD5 hash digests to plaintext passwords.

- **Model**: Transformer encoder-decoder architecture
- **Task**: Sequence-to-sequence learning (hash → password)
- **Dataset**: 1M hash-password pairs
- **Features**: TensorBoard logging, checkpoint saving/loading, evaluation metrics

## 1. Imports

Import all necessary libraries and modules.

````python
import sys
sys.dont_write_bytecode = True
````

````python
# imports
import torch
from torch.utils.data import DataLoader
from pathlib import Path
# local
from model import OptimusPrime
from optimizer import AdamWarlock
from data import Bumblebee, collate_batch, ALLOWED_PW_CHARS as PW_VOCAB, PAD_ID, SOS_ID, EOS_ID
from trainer import Trainer
````

## 2. Configuration

`Hyperparameters`

````python
# model
EPOCHS = 10
TRAIN_BATCH_SIZE = 256 # 128, 256, 512
EVAL_BATCH_SIZE = 128
LR = 3e-5  # base learning rate

# trainer
WEIGHT_LOAD_MODE = 'latest'  # 'best', 'latest', or 'none'
DEVICE = 'cuda' if torch.cuda.is_available() else 'cpu'

# eval
EVAL_TEMPERATURE = 1.2
REPITION_PENALTY = 1.5

# data sampling - Sweet spot: not too big, not too small
TRAIN_SAMPLE_FRACTION = 0.30  # 30% = 3M samples
TRAIN_SAMPLE_SEED = None      # set int for reproducible sampling

# paths
TRAIN_DATA_PATH = Path.cwd().parent / 'data' / 'training' / '10MIL_train.tsv'
EVAL_DATA_PATH = Path.cwd().parent / 'data' / 'eval' / '10K_eval.tsv'
````

`Datasets & Dataloaders`

````python
# datasets - initial load
TRAIN_DATA = Bumblebee(TRAIN_DATA_PATH, sample_fraction = TRAIN_SAMPLE_FRACTION, sample_seed = TRAIN_SAMPLE_SEED)
EVAL_DATA = Bumblebee(EVAL_DATA_PATH)   # optional: for automatic eval during training

# dataloader kwargs (reusable)
TRAIN_LOADER_KWARGS = dict(batch_size = TRAIN_BATCH_SIZE, shuffle = True, collate_fn = collate_batch, num_workers = 4, pin_memory = True)
EVAL_LOADER_KWARGS = dict(batch_size = EVAL_BATCH_SIZE if not isinstance(EVAL_BATCH_SIZE, type(...)) else 128, collate_fn = collate_batch)

# initial dataloaders
t_dataloader = DataLoader(TRAIN_DATA, **TRAIN_LOADER_KWARGS)
e_dataloader = DataLoader(EVAL_DATA, **EVAL_LOADER_KWARGS)
````

````python
print(f'[training file]: {TRAIN_DATA_PATH.name}')
print(f'[subset]: {f"{TRAIN_SAMPLE_FRACTION:.0%}" if TRAIN_SAMPLE_FRACTION else "100%"}')
print(f'[examples loaded]: {len(TRAIN_DATA):,}')
print(f'[batches per epoch]: {len(t_dataloader):,}')
````

## 3. Models

`OptimusPrime`  
transformer encoder-decoder model

````python
# create model
model = OptimusPrime(
    vocab_size = 256,
    pw_vocab_size = len(PW_VOCAB),
    pad_id = PAD_ID,
    sos_id = SOS_ID,
    eos_id = EOS_ID,
    d_model = 512,  # (2x) increases the representation capacity of each token embedding
    n_heads = 8,    # (same) allows model to attend to different aspects simultaneously
    num_layers = 6, # (+2 layers) adds more encoder/decoder layers for hierarchal feature learning
    ff_dim = 2048,  # (4x d_model) expands the MLP layer w/in each transformer block
    dropout = 0.1,
    label_smoothing = 0.06     # prevent overconfidence (.05 - .1)
)
````

````python
# count parameters
total_params = sum(p.numel() for p in model.parameters())
trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)

print(f'Total parameters: {total_params:,}')
print(f'Trainable parameters: {trainable_params:,}')
````

`AdamWarlock`  
AdamW optimizer wrapper

````python
# create optimizer with scheduling
optimizer = AdamWarlock(
    model.parameters(),
    lr = LR,
    t_max = EPOCHS * len(t_dataloader),
    # total_steps = len(t_dataloader),  # previous arg
    # schedule = "cosine",             # previous arg
    eta_min = 1e-6,
    weight_decay = 0.01,
)
````

`Trainer`

````python
# trainer kwargs (reusable)
TRAINER_KWARGS = {
    'model': model,
    'optimizer': optimizer,
    'dataloader': t_dataloader,
    'eval_dataloader': e_dataloader,
    'device': DEVICE,
    'epochs': EPOCHS,
    'checkpoint_dir': Path('checkpoints'),
    'checkpoint_interval': 1,
    'logs': 'runs/optimus3.0.2',
    'keep': True,
    'load_mode': WEIGHT_LOAD_MODE,
    'eval_checkpoint': 5,
    'max_checkpoints': 5,
    'temperature': EVAL_TEMPERATURE,
    'repetition_penalty': REPITION_PENALTY,
    'accumulation_steps': 4,
}

# initial trainer
trainer = Trainer(**TRAINER_KWARGS)
````

## 4. Train

Run full training loop with gradient tracking, TensorBoard logging, and checkpoint saving.   

Each epoch trains on a random 10% slice of the consolidated dataset. Recreate `TRAIN_DATA` if you want a fresh draw between runs.

````python
# setup trainer / load weights
trainer.load()

START = trainer.start_epoch
END = trainer.start_epoch + EPOCHS

# training loop - fresh dataset, dataloader, and trainer each epoch
for epoch in range(START, END):
    # reload everything fresh (skip first epoch - already loaded)
    if epoch > START:
        # fresh dataset
        TRAIN_DATA = Bumblebee(TRAIN_DATA_PATH, sample_fraction = TRAIN_SAMPLE_FRACTION, sample_seed = TRAIN_SAMPLE_SEED)
        
        # fresh dataloader
        t_dataloader = DataLoader(TRAIN_DATA, **TRAIN_LOADER_KWARGS)
        
        # fresh trainer with updated dataloader
        TRAINER_KWARGS['dataloader'] = t_dataloader
        trainer = Trainer(**TRAINER_KWARGS)
        
        # reload weights into fresh trainer
        trainer.start_epoch = epoch
        trainer.load('latest')

    # train one epoch
    epoch_loss, global_step = trainer.train(epoch, start = START, end = END)

    # save best FIRST (so best_loss gets updated before regular checkpoint)
    if epoch_loss < trainer.best_loss:
        trainer.save(epoch, epoch_loss)

    # save checkpoint AFTER (so it has the updated best_loss)
    if (epoch + 1) % trainer.checkpoint_interval == 0:
        trainer.save(epoch, epoch_loss, global_step)

    # eval
    if trainer.eval_dataloader is not None and (epoch + 1) % trainer.eval_checkpoint == 0:
        print(f'\n[eval]: epoch {epoch + 1}')
        trainer.eval(trainer.eval_dataloader, step = epoch + 1)

    trainer.start_epoch = epoch
````
